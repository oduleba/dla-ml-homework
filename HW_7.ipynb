```python
import numpy as np
```

```python
#Завдання 1. Створення масивів

mylist = np.array([5, 15, 25, 35, 45])
zeros = np.zeros((7))
ones = np.ones((6))
new_list = np.arange(10,20)

print (mylist)
print (zeros)
print (ones)
print (new_list)
```

```python
#Завдання 2. Базові операції

arr_new = np.array([12, 24, 36, 48, 60])

arr_1 = arr_new*2
arr_2 = arr_new + 10
arr_3 = arr_new**2
arr_4 = np.sqrt(arr_new)

print (arr_1)
print (arr_2)
print (arr_3)
print (arr_4)
```

```python
#Завдання 3. Статистичні функції

data = np.array([18, 27, 33, 41, 52, 68, 79])

print(np.mean(data))
print(np.min(data))
print(np.max(data))
print(np.sum(data))
```

```python
#Завдання 4. Індексація

arr = np.array([15, 22, 35, 44, 58, 67])
arr_more40 = arr>40

print(arr[0])
print(arr[-1])
print(arr[1:4])
print(arr > 40)
print(arr[arr_more40])
```

```python
#Завдання 5. Двовимірні масиви

matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14,16,18]])
print(matrix.shape)
print(matrix[0])
print(matrix[:,1])
print(np.sum(matrix))
print(np.mean(matrix))
```
