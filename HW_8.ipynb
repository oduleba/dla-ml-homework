```python
#Завдання 1. Завантаження та огляд

import pandas as pd

df = pd.read_csv('tips.csv')

display(df.head())
display(df.shape)
display(df.columns)
display(df.info)
```

```python
#Завдання 2. Базова вибірка даних

total_bill = df['total_bill']
bill_and_tip = df[['total_bill', 'tip']]
#top_rows = df.head(10)
top_rows = df.head(10)
row_5 = df.iloc[5]

display(total_bill)
display(bill_and_tip)
display(top_rows)
display(row_5)
```

```python
from IPython.core.interactiveshell import dis
#Завдання 3. Фільтрація даних

bill_over20 = df[df['total_bill'] > 20]
female_visitors = df[df['sex'] == 'Female']
dinner_bill = df[df['time'] == 'Dinner']
dinner_bill_over30 = df[(df['total_bill']>30) & (df['time'] == 'Dinner')]

print(f'Рахунки більше 20 доларів. Кількість записів: {len(bill_over20)}')
display(bill_over20.head(5))

print(f'Всі відвідувачі-жінки. Кількість записів: {len(female_visitors)}')
display(female_visitors.head(5))

print(f'Рахунки на вечерю. Кількість записів: {len(dinner_bill)}')
display(dinner_bill.head(5))

print(f'Рахунки на вечерю з сумою чеку більше 30 доларів. Кількість записів: {len(dinner_bill_over30)}')
display(dinner_bill_over30.head(5))
```

```python
from numpy import mean
#Завдання 4. Описова статистика

average_bill = df['total_bill'].mean().round(2)
print (f'Середня розмір рахунку: {average_bill}')

max_tip = df['tip'].max()
print (f'Максимальні чайові: {max_tip}')

null_count = df.isnull().sum().sum()
print (f'Кількість пропущенних значень: {null_count}')

dinner = df['time'].value_counts()['Dinner']
lunch = df['time'].value_counts()['Lunch']
print(f'Кількість рахунків за обіди: {lunch}, кількість рахунків за вечері: {dinner}')
```

```python
#Завдання 5. Статистика із групуванням

by_sex = df.groupby('sex')['total_bill'].mean().round(2)
print (by_sex)

tips_by_day = df.groupby('day')['tip'].mean().round(2)
print (tips_by_day)

visitors_by_day = df.groupby('day')['size'].sum()
print(visitors_by_day)
```

```python
#Завдання 6. Створення нової колонки

df['tip_percentage'] = (df['tip'] / df['total_bill'] * 100).round(2)
display(df.head(10))
```

```python
#Завдання 7. Сортування

df_sorted_bill = df.sort_values('total_bill')
df_sorted_tip = df.sort_values('tip', ascending = False)
top_tip = df.nlargest(5, 'tip')

print("--- Сортування за розміром рахунку (зростання) ---")
display(df_sorted_bill.head(10))

print("\n--- Сортування за чайовими (спадання) ---")
display(df_sorted_tip.head(10))

print("\n--- Топ-5 найбільших чайових (через nlargest) ---")
display(top_tip)
```
