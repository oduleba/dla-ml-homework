```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt

df = pd.read_csv('tips.csv')
```

```python
#Завдання 1. Гістограма розподілу рахунків
df['total_bill'].hist(bins=20, color = 'teal', edgecolor='black')
plt.title('Загальний рахунок')
plt.xlabel('Сума')
plt.ylabel('Кількість')
plt.show()
```

```python
#Завдання 2. Box plot чайових за днями тижня
day_order = ['Thur', 'Fri', 'Sat', 'Sun']
df['day'] = pd.Categorical(df['day'], categories=day_order, ordered=True)


df.boxplot(column='tip', by='day', figsize=(7, 5))
plt.title('Чайові за днями тижня')
plt.suptitle('')
plt.xlabel('День тижня')
plt.ylabel('Розмір Чайових')
plt.show()

max_median_tip_day = df.groupby('day')['tip'].median().idxmax()
print(f"Найвище медіанне значення чайових у {max_median_tip_day}")
```

```python
#Завдання 3. Scatter plot: рахунок vs чайові

df.plot.scatter(x='total_bill', y='tip', alpha=0.5, c='size', colormap='plasma')

plt.title('Залежність між рахунком та чайовими')
plt.xlabel('Рахунок')
plt.ylabel('Чайові')
plt.show();
```

```python
#Завдання 4. Bar chart: середній рахунок за часом

df.groupby('time')['total_bill'].mean().plot(kind='bar', rot = 0, color='bisque', edgecolor='black')
plt.title('Середній рахунок за часом відвідування')
plt.xlabel('Час відвідування')
plt.ylabel('Рахунок')
plt.show()
```

```python
#Завдання 5. Інтерактивна діаграма розсіювання
#Створіть з допомогою бібліотек Plotly інтерактивний scatter plot для total_bill vs tip з кольором за статтю (sex).

import plotly.express as px
import plotly.graph_objects as go

fig = px.scatter(df, x='total_bill', y='tip', color='sex',
                 hover_data=['smoker', 'size', 'time'],
                 title='Рахунок vs Чайові',
                 labels={'sex': 'Стать', 'tip': 'Чайові', 'total_bill': 'Рахунок', 'smoker': 'Курці', 'size': 'Кількість гостей', 'time': 'Час відвідування'},
                 color_discrete_map = {'Male' : 'blue', 'Female' : 'red'})
fig.show()
```

```python
#Завдання 6. Інтерактивний box plot: чайові за днями та статтю
#Створіть інтерактивний box plot для чайових за днями тижня з розбивкою за кольором по статі.

fig = px.box(df, x='day', y='tip', color='sex',
             title='Розподіл чайових за днями та статтю',
             labels={'day': 'День тижня', 'tip': 'Чайові', 'sex': 'Стать'})
fig.show()
```
