```python
#Завдання 2. Цикл for по списку
```

```python
names = ['Олег', 'Анна', 'Іван', 'Марія']

for name in names:
    print (f'Привіт, {name}!')
```

```python
#Завдання 3. Цикл з enumerate
```

```python
grades = [85, 92, 78, 95, 88]

indeces = []
for i, grade in enumerate(grades):
    print (f'Студент {i+1}: {grade} балів')
    indeces.append(i)
```

```python
#Завдання 4. Сума парних чисел
```

```python
Sum = 0
for i in range(1,21):
    if i % 2 ==0:
        Sum += i
print (Sum)
```
